### cosine_similarity

```python
# Cosine Similarity 1
import numpy as np

# define the vectors
a = np.array([-0.4, 0.8])
b = np.array([-0.3, 0.2])

# dot product
dot_product = np.dot(a, b)

# magnitudes (norms)
norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

# cosine similarity
cos_similarity = dot_product / (norm_a * norm_b)

print(f'Cosine similarity: {cos_similarity:.4f}')
```

```python
# Cosine Similarity 2 
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

a = np.array([[-0.4, 0.8]])
b = np.array([[-0.3, 0.2]])

similarity = cosine_similarity(a, b)
print(f'Cosine similarity: {similarity[0][0]:.4f}')
```

### Enumerate

```python
fruits = ['apple', 'banana', 'cherry']
for i, name in enumerate(fruits, start=1):
    print(i, name)
```
