# Language Modeling and Perplexity: Comprehensive Guide

This notebook explores Language Modeling, the Sparsity Problem, and the evaluation metric: **Perplexity**.

## 1. Evaluation: Perplexity

Perplexity measures how 'surprised' a model is by a sequence of data. Lower values indicate a better model.

### Formula
$$PP(W) = \exp\left(-\frac{1}{N} \sum_{i=1}^{N} \log P(w_i \mid w_1,\ldots,w_{i-1})\right)$$

## 2. The Sparsity Problem

When training on finite corpora, many perfectly valid sequences may never appear. 
* **Problem:** MLE (counting) assigns 0 probability to these sequences.
* **Fix:** Smoothing (Add-k, Interpolation, Backoff).

## 3. Smoothing Techniques

* **Add-k Smoothing ($0 < k < 1$):** Adds a small constant to all counts to avoid zero probability.
* **Interpolation:** $P_{\text{interp}} = \lambda_1 P_{3} + \lambda_2 P_{2} + \lambda_3 P_{1}$.
* **Backoff:** If a higher-order N-gram is missing, fall back to a lower-order model.

```python
# Example: Trigram Model with Add-k Smoothing Setup
import torch

# Assuming vocab_size is known (e.g., 28)
vocab_size = 28
k = 0.01
# Initialize with Add-k
N = torch.full((vocab_size, vocab_size, vocab_size), k, dtype=torch.float32)

# In practice, you would iterate through your training data and increment N
# N[ix1, ix2, ix3] += 1

# Normalize to probabilities
P = N / N.sum(dim=2, keepdim=True)
print('Matrix P initialized and normalized.')
```
